## Setup

```python
from pathlib import Path

import joblib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import GradientBoostingRegressor, RandomForestRegressor
from sklearn.impute import SimpleImputer
from sklearn.inspection import DecisionBoundaryDisplay, permutation_importance
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, mean_absolute_error, mean_squared_error, roc_auc_score
from sklearn.model_selection import (
    GridSearchCV, KFold, StratifiedKFold, cross_val_score, train_test_split,
)
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier, export_text, plot_tree


def find_data_dir(name="all_datasets", marker="titanic_dataset.csv"):
    """Search the script/working folder and their parents for a data folder holding `marker`."""
    starts = [Path.cwd()]
    try:
        starts.insert(0, Path(__file__).resolve().parent)  # not defined inside a notebook
    except NameError:
        pass
    for start in starts:
        for folder in [start, *start.parents]:
            if (folder / name / marker).is_file():
                return folder / name
    searched = ", ".join(str(s) for s in starts)
    raise FileNotFoundError(f"Could not find {name}/{marker} in or above: {searched}")


# The lab's original `Path.cwd().parent` only works for one folder depth; searching upward works for any depth.
DATA = find_data_dir()
PROJECT_ROOT = DATA.parent
MODELS = PROJECT_ROOT / "models"
MODELS.mkdir(exist_ok=True)
print("Using data from:", DATA)
print("Saving models to:", MODELS)
```

## Part A — KNN versus Gaussian naive Bayes

KNN votes among standardized neighbours; Gaussian naive Bayes combines per-feature Gaussian likelihoods and class priors.

```python
cancer = pd.read_csv(DATA / "breast_cancer_dataset.csv")
X_c = cancer.drop(columns=["id", "diagnosis", "Unnamed: 32"], errors="ignore")
y_c = (cancer["diagnosis"] == "M").astype(int)

print(X_c)

Xc_train, Xc_valid, yc_train, yc_valid = train_test_split(
    X_c, y_c, test_size=0.25, random_state=42, stratify=y_c
)
print("features:", X_c.shape[1], " train:", len(Xc_train), " validation:", len(Xc_valid),
      " malignant share:", round(y_c.mean(), 3))
```

`k` is a hyperparameter, so it is chosen by cross-validation **inside the training partition**. Small `k` follows every local quirk (low bias, high variance); large `k` smooths everything (high bias, low variance).

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
k_values = [1, 3, 5, 7, 11, 21]
cv_scores = {}
for k in k_values:
    candidate = Pipeline([
        ("scale", StandardScaler()),
        ("model", KNeighborsClassifier(n_neighbors=k, weights="distance")),
    ])
    scores = cross_val_score(candidate, Xc_train, yc_train, cv=cv, scoring="roc_auc")
    cv_scores[k] = scores.mean()
    print(f"k={k:2d}  cross-validated ROC AUC = {scores.mean():.3f}")
best_k = max(cv_scores, key=cv_scores.get)
print("selected k:", best_k)

plt.figure(figsize=(5, 3.2))
plt.plot(k_values, [cv_scores[k] for k in k_values], marker="o")
plt.axvline(best_k, linestyle="--", color="grey")
plt.xlabel("k (number of neighbours)")
plt.ylabel("Cross-validated ROC AUC")
plt.title("Choosing k on training data only")
plt.tight_layout()
plt.show()
```

**Inspect.** Cross-validated AUC jumps from 0.946 at `k=1` to 0.984 at `k=3`, then creeps up to 0.987 (`k=7`) and 0.988 (`k=11` and `k=21`). So validation improves beyond `k=1` and is flat for `k=7–21`: the differences there are 0.001, far below what five folds can resolve. The code picks `k=11` because it is the first maximum at full precision, but `k=7` or `k=21` would be just as defensible. The single-neighbour model is the one that "follows every local quirk": one mislabelled or unusual training tumour flips its neighbours' votes.

With `k` locked, both models are compared on the validation split.

```python
models = {
    "KNN": Pipeline([
        ("scale", StandardScaler()),
        ("model", KNeighborsClassifier(n_neighbors=best_k, weights="distance")),
    ]),
    "GaussianNB": Pipeline([
        ("scale", StandardScaler()),
        ("model", GaussianNB()),
    ]),
}
for name, model in models.items():
    model.fit(Xc_train, yc_train)
    pred = model.predict(Xc_valid)
    prob_c = model.predict_proba(Xc_valid)[:, 1]
    print(f"{name:10s} accuracy={accuracy_score(yc_valid, pred):.3f} "
          f"roc_auc={roc_auc_score(yc_valid, prob_c):.3f}")
```

**Inspect.** The two models split the metrics. KNN has the higher accuracy (0.958, which is 137 of 143 validation tumours, against 0.944 or 135 of 143), while Gaussian naive Bayes has the higher ROC AUC (0.992 against 0.986). That is a two-patient difference in accuracy on one split of 143, so neither model has shown itself better; both are excellent on all 30 features. The accuracy and AUC rankings can disagree because accuracy only looks at the 0.5 threshold, while AUC scores how well the whole ranking separates malignant from benign.

Compare two-feature boundaries on tumour radius and texture, adding logistic regression as a reference.

```python
two = ["radius_mean", "texture_mean"]
fig, axes = plt.subplots(1, 3, figsize=(12, 3.8))
for ax, (name, estimator) in zip(axes, [
    ("Logistic regression", LogisticRegression(max_iter=2000)),
    (f"KNN (k={best_k})", KNeighborsClassifier(n_neighbors=best_k, weights="distance")),
    ("Gaussian naive Bayes", GaussianNB()),
]):
    small = Pipeline([("scale", StandardScaler()), ("model", estimator)])
    small.fit(Xc_train[two], yc_train)
    DecisionBoundaryDisplay.from_estimator(small, Xc_valid[two], ax=ax, alpha=0.3,
                                           response_method="predict", cmap="coolwarm")
    ax.scatter(Xc_valid[two[0]], Xc_valid[two[1]], c=yc_valid, cmap="coolwarm",
               s=12, edgecolor="k", linewidth=0.3)
    acc = accuracy_score(yc_valid, small.predict(Xc_valid[two]))
    print(f"{name:22s} two-feature validation accuracy = {acc:.3f}")
    ax.set_title(f"{name}: {acc:.2f} accuracy")
    ax.set_xlabel("radius_mean")
    ax.set_ylabel("texture_mean")
plt.tight_layout()
plt.show()
```

**Inspect.** With only radius and texture, validation accuracy is 0.881 for logistic regression, 0.895 for KNN and 0.860 for Gaussian naive Bayes, all below the 0.94–0.96 that KNN and naive Bayes reach with 30 features, because two columns cannot carry everything. The shapes follow from the model families: logistic regression can only draw a straight line, naive Bayes (which fits a separate variance per class) draws a smooth curved boundary, and KNN's boundary follows the local mix of neighbours and so is the most irregular. The accuracy gaps are 2–5 tumours out of 143, so the shapes matter more here than the ranking.

## Part B — One readable decision tree, and what happens as it grows

Trees split on value order, so numerical preprocessing imputes missingness without scaling.

```python
titanic = pd.read_csv(DATA / "titanic_dataset.csv")
features = ["Pclass", "Sex", "Age", "SibSp", "Parch", "Fare", "Embarked"]
t_num = ["Age", "SibSp", "Parch", "Fare"]
t_cat = ["Pclass", "Sex", "Embarked"]
X_t = titanic[features]
y_t = titanic["Survived"]

tree_preprocess = ColumnTransformer([
    ("num", SimpleImputer(strategy="median"), t_num),
    ("cat", Pipeline([
        ("impute", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore")),
    ]), t_cat),
])
Xt_train, Xt_valid, yt_train, yt_valid = train_test_split(
    X_t, y_t, test_size=0.25, random_state=42, stratify=y_t
)
tree = Pipeline([
    ("preprocess", tree_preprocess),
    ("tree", DecisionTreeClassifier(max_depth=4, min_samples_leaf=12, random_state=42)),
])
tree.fit(Xt_train, yt_train)
print("training accuracy:", round(accuracy_score(yt_train, tree.predict(Xt_train)), 3))
print("validation accuracy:", round(accuracy_score(yt_valid, tree.predict(Xt_valid)), 3))
```

Print and plot the tree; inspect split questions, sample counts, and classes.

```python
feature_names = list(tree.named_steps["preprocess"].get_feature_names_out())
print(export_text(tree.named_steps["tree"], feature_names=feature_names, max_depth=3))

plt.figure(figsize=(12, 6))
plot_tree(tree.named_steps["tree"], feature_names=feature_names,
          class_names=["did not survive", "survived"], max_depth=2, filled=True, fontsize=8)
plt.title("Top of the fitted Titanic decision tree")
plt.tight_layout()
plt.show()
```

**Inspect.** The first question is sex, as the lab says. For men the next split is age (`Age <= 3.5`: young boys are predicted to survive) and then first class versus the rest; for women it is class (`Pclass_3`), then fare and age. A telling detail: third-class women with `Fare > 7.89` and `Age > 5.5` are the one female branch predicted *not* to survive. This tree's training accuracy is 0.843 and its validation accuracy 0.767, so it already shows a 7.6-point gap even though it is only depth 4. (`export_text` was told `max_depth=3`, so the printout stops before the last split level and every leaf shows only its majority class.)

Fit each depth and compare training with validation accuracy.

```python
depths = list(range(1, 13))
train_acc, valid_acc = [], []
for depth in depths:
    t = Pipeline([
        ("preprocess", tree_preprocess),
        ("tree", DecisionTreeClassifier(max_depth=depth, random_state=42)),
    ]).fit(Xt_train, yt_train)
    train_acc.append(accuracy_score(yt_train, t.predict(Xt_train)))
    valid_acc.append(accuracy_score(yt_valid, t.predict(Xt_valid)))

plt.figure(figsize=(6, 3.6))
plt.plot(depths, train_acc, marker="o", label="training")
plt.plot(depths, valid_acc, marker="o", label="validation")
plt.xlabel("max_depth")
plt.ylabel("Accuracy")
plt.title("Deeper trees memorise; validation accuracy stops improving")
plt.legend()
plt.tight_layout()
plt.show()

print(pd.DataFrame({"depth": depths, "training": train_acc, "validation": valid_acc})
      .round(3).to_string(index=False))
```

**Inspect.** Training accuracy climbs steadily from 0.793 at depth 1 to 0.954 at depth 12, while validation accuracy peaks at depth 3 (0.785) and ends at 0.731. That is where overfitting shows: past depth 3 the extra splits fit training passengers and stop helping unseen ones. A caution on resolution: the validation set has 223 passengers, so one passenger is 0.45 points, and depths 1–7 (0.758–0.785) are not clearly different from each other. What is clear is the trend: every depth of 8 or more (0.749 or lower) is worse than every depth of 3 or less.

## Part C — Random forest versus gradient boosting

Model `log(1 + price)` from Ames Housing's mixed-type features. Compare random forest and gradient boosting on identical three-fold splits.

```python
ames = pd.read_csv(DATA / "ames_housing_dataset.csv")
X_a = ames.drop(columns=["SalePrice", "Order", "PID"], errors="ignore")  # identifiers out
y_a = np.log1p(ames["SalePrice"])
print("Ames:", X_a.shape, " numeric:", X_a.select_dtypes(include="number").shape[1],
      " categorical:", X_a.select_dtypes(exclude="number").shape[1])

ames_preprocess = ColumnTransformer([
    ("num", SimpleImputer(strategy="median"), X_a.select_dtypes(include="number").columns),
    ("cat", Pipeline([
        ("impute", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore", sparse_output=False)),
    ]), X_a.select_dtypes(exclude="number").columns),
])
```

```python
ensembles = {
    "random_forest": RandomForestRegressor(
        n_estimators=150, max_features=0.8, min_samples_leaf=2, random_state=42, n_jobs=-1
    ),
    "gradient_boosting": GradientBoostingRegressor(
        n_estimators=150, learning_rate=0.05, max_depth=3, loss="huber", random_state=42
    ),
}
cv3 = KFold(n_splits=3, shuffle=True, random_state=42)
cv_rmse = {}
for name, estimator in ensembles.items():
    pipe = Pipeline([("preprocess", ames_preprocess), ("model", estimator)])
    neg_mse = cross_val_score(pipe, X_a, y_a, cv=cv3, scoring="neg_mean_squared_error")
    cv_rmse[name] = np.sqrt(-neg_mse)
    print(f"{name:18s} CV log-RMSE = {cv_rmse[name].mean():.4f} +/- {cv_rmse[name].std():.4f}")
print("per-fold:", {k: v.round(4).tolist() for k, v in cv_rmse.items()})
gap = cv_rmse["random_forest"] - cv_rmse["gradient_boosting"]
print("per-fold gap (forest minus boosting):", gap.round(4).tolist(),
      f" mean = {gap.mean():.4f}, boosting ahead in {(gap > 0).sum()} of {len(gap)} folds")

plt.figure(figsize=(5, 3.4))
for i, (name, values) in enumerate(cv_rmse.items()):
    plt.scatter([i] * len(values), values, s=60, label="fold scores" if i == 0 else None)
    plt.hlines(values.mean(), i - 0.2, i + 0.2, color="black", label="mean" if i == 0 else None)
plt.xticks(range(len(cv_rmse)), list(cv_rmse))
plt.ylabel("log-RMSE per fold (lower is better)")
plt.title("Fold-by-fold: is the gap bigger than the spread?")
plt.legend()
plt.tight_layout()
plt.show()
```

**Inspect.** Gradient boosting has the lower error on average: log-RMSE 0.1273 against 0.1382 for the forest (roughly a 13% against 14% typical price error). Taken against each model's fold spread (about 0.012–0.014), the 0.011 gap looks modest, as the lab says. But the folds are the same for both models and some folds are simply harder (fold 2 is the worst for both), so the paired view matters: boosting beat the forest in all 3 folds, by 0.007 to 0.013. That is a consistent edge, but three folds is thin evidence, and boosting has more settings to get right (learning rate, depth, loss), so the gain is not free. I did not time the two models, so this notebook cannot say how runtime compares.

Repeat on eleven wine measurements. Permute validation columns to measure the fitted forest's reliance on each.

```python
wine = pd.read_csv(DATA / "red_wine_quality_dataset.csv")
X_w = wine.drop(columns="quality")
y_w = wine["quality"]
Xw_train, Xw_valid, yw_train, yw_valid = train_test_split(X_w, y_w, test_size=0.25, random_state=42)

wine_models = {
    "random_forest": RandomForestRegressor(n_estimators=400, min_samples_leaf=2,
                                           random_state=42, n_jobs=-1),
    "gradient_boosting": GradientBoostingRegressor(random_state=42),
}
wine_pred = {}
for name, model in wine_models.items():
    model.fit(Xw_train, yw_train)
    wine_pred[name] = model.predict(Xw_valid)
    print(f"{name:18s} MAE = {mean_absolute_error(yw_valid, wine_pred[name]):.3f} "
          f"RMSE = {np.sqrt(mean_squared_error(yw_valid, wine_pred[name])):.3f}")
```

```python
importance = permutation_importance(
    wine_models["random_forest"], Xw_valid, yw_valid, n_repeats=10, random_state=42, n_jobs=-1
)
ranked = pd.Series(importance.importances_mean, index=X_w.columns).sort_values()
spread = importance.importances_std[np.argsort(importance.importances_mean)]

fig, axes = plt.subplots(1, 2, figsize=(11, 3.8))
axes[0].barh(ranked.index, ranked.values, xerr=spread)
axes[0].set_xlabel("Increase in error when the column is shuffled")
axes[0].set_title("Permutation importance (random forest)")
axes[1].scatter(yw_valid, wine_pred["gradient_boosting"], alpha=0.5)
axes[1].plot([3, 8], [3, 8], linestyle="--", color="grey")
axes[1].set_xlabel("Observed quality")
axes[1].set_ylabel("Predicted quality")
axes[1].set_title("Gradient boosting pulls extremes to the middle")
plt.tight_layout()
plt.show()

print(ranked.sort_values(ascending=False).round(3).to_string())
print("\nobserved quality range:", yw_valid.min(), "-", yw_valid.max(),
      " | boosting prediction range:", wine_pred["gradient_boosting"].min().round(2), "-",
      wine_pred["gradient_boosting"].max().round(2))
```

**Inspect.** Here the forest wins: MAE 0.432 against 0.476 and RMSE 0.565 against 0.597, the reverse of Ames. One caveat on that comparison: the Ames boosting model had hand-set hyperparameters, while the wine boosting model uses scikit-learn's defaults on a single split, so this is not a like-for-like contest and "forests beat boosting on wine" is a statement about these settings. Shuffling `alcohol` raises the forest's error by 0.380, `sulphates` by 0.224 and `volatile acidity` by 0.117; every other column is 0.053 or less, so a handful of columns carry most of the signal. Boosting's predictions run from 3.52 to 7.41 while the observed ratings span 3 to 8, so the rare extreme wines are pulled toward the middle. Permutation importance can also understate columns that are correlated with others (acidity, density and alcohol overlap), because the model can lean on a correlated partner when one is shuffled.

## Part D — Support vector machines: two dials and a kernel

Fit the SVM with scaling inside each training fold. `C` controls violation penalties; the kernel controls boundary shape; RBF `gamma` controls locality.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_c, y_c, test_size=0.25, random_state=42, stratify=y_c
)
svm_pipe = Pipeline([("scale", StandardScaler()), ("model", SVC())])
search = GridSearchCV(
    svm_pipe,
    param_grid={
        "model__kernel": ["linear", "rbf"],
        "model__C": [0.1, 1, 10, 100],
        "model__gamma": ["scale", 0.01, 0.1],
    },
    cv=StratifiedKFold(5, shuffle=True, random_state=42),
    scoring="roc_auc",
    n_jobs=-1,
)
search.fit(X_train, y_train)
score = search.decision_function(X_test)
pred = search.predict(X_test)
print("best parameters:", search.best_params_)
print("cross-validated ROC AUC:", round(search.best_score_, 3))
print("test accuracy:", round(accuracy_score(y_test, pred), 3))
print("test ROC AUC:", round(roc_auc_score(y_test, score), 3))
```

**Inspect.** The search picks the RBF kernel with `C=10` and `gamma=0.01`: cross-validated AUC 0.995, test accuracy 0.979 (140 of 143) and test AUC 0.997. The test set played no part in choosing these settings, so those numbers are an honest read of the chosen model. Note that `C=10` and `C=100` tie at `gamma=0.01` to three decimals (0.995 each), so the choice between them is not meaningful.

Plot cross-validated AUC for each (`C`, `gamma`) pair and compare two-feature kernel boundaries.

```python
results = pd.DataFrame(search.cv_results_)
rbf = results[results["param_model__kernel"] == "rbf"].copy()
rbf["gamma"] = rbf["param_model__gamma"].astype(str)
grid = rbf.pivot(index="param_model__C", columns="gamma", values="mean_test_score")

fig, axes = plt.subplots(1, 3, figsize=(13, 3.8))
im = axes[0].imshow(grid.values, cmap="viridis", aspect="auto")
axes[0].set_xticks(range(len(grid.columns)), grid.columns)
axes[0].set_yticks(range(len(grid.index)), grid.index)
axes[0].set_xlabel("gamma")
axes[0].set_ylabel("C")
for i in range(grid.shape[0]):
    for j in range(grid.shape[1]):
        axes[0].text(j, i, f"{grid.values[i, j]:.3f}", ha="center", va="center",
                     color="white", fontsize=8)
axes[0].set_title("RBF kernel: CV ROC AUC by C and gamma")

for ax, (title, estimator) in zip(axes[1:], [
    ("Linear kernel", SVC(kernel="linear", C=1)),
    ("RBF kernel (gamma = 1)", SVC(kernel="rbf", C=1, gamma=1.0)),
]):
    small = Pipeline([("scale", StandardScaler()), ("model", estimator)]).fit(X_train[two], y_train)
    DecisionBoundaryDisplay.from_estimator(small, X_test[two], ax=ax, alpha=0.3,
                                           response_method="predict", cmap="coolwarm")
    ax.scatter(X_test[two[0]], X_test[two[1]], c=y_test, cmap="coolwarm",
               s=12, edgecolor="k", linewidth=0.3)
    acc = accuracy_score(y_test, small.predict(X_test[two]))
    print(f"{title:24s} two-feature test accuracy = {acc:.3f}")
    ax.set_title(f"{title}: {acc:.2f} accuracy")
    ax.set_xlabel("radius_mean")
    ax.set_ylabel("texture_mean")
plt.tight_layout()
plt.show()

print("RBF grid (rows = C, columns = gamma):")
print(grid.round(3).to_string())
print("\nbest linear-kernel CV AUC:",
      round(results.loc[results["param_model__kernel"] == "linear", "mean_test_score"].max(), 3))
```

**Inspect.** Most scores are similar: the RBF grid runs only from 0.978 to 0.995, and the best linear kernel reaches 0.993 in cross-validation. The dials do matter at the edges. `gamma=0.1` is the weakest column at every `C`, and the gap widens at larger `C` (0.982 and 0.981 at `C=10` and `C=100`, against 0.995 for `gamma=0.01`): a more local kernel with a larger penalty starts chasing individual training points. A very small penalty (`C=0.1`) is also a little worse (0.978–0.985), a smoother, more tolerant boundary. In the two-feature panels, the linear kernel scores 0.881 and the deliberately flexible RBF (`gamma=1`, ten times the grid's largest value) 0.909, a gap of 4 tumours out of 143. With two features there is something for curvature to fit; with all 30 scaled features the classes are close to linearly separable, so RBF gains very little.

**1. Grow the Titanic tree with `max_depth=None` and `min_samples_leaf=1`, and compare training and validation accuracy with the depth-4 tree.**

```python
def titanic_tree(**params):
    return Pipeline([
        ("preprocess", tree_preprocess),
        ("tree", DecisionTreeClassifier(random_state=42, **params)),
    ]).fit(Xt_train, yt_train)


tree_variants = {
    "depth 4, min_samples_leaf=12": titanic_tree(max_depth=4, min_samples_leaf=12),
    "unrestricted (None, 1)": titanic_tree(max_depth=None, min_samples_leaf=1),
}
rows = []
for name, fitted in tree_variants.items():
    t = fitted.named_steps["tree"]
    rows.append({
        "tree": name,
        "depth": t.get_depth(),
        "leaves": t.get_n_leaves(),
        "training accuracy": accuracy_score(yt_train, fitted.predict(Xt_train)),
        "validation accuracy": accuracy_score(yt_valid, fitted.predict(Xt_valid)),
    })
print(pd.DataFrame(rows).set_index("tree").round(3).to_string())
```

**Inspect.** The unrestricted tree grows to depth 21 with 146 leaves and reaches 0.978 training accuracy (not exactly 1.0, most likely because some passengers have identical feature values but different outcomes, which no tree can separate). Its validation accuracy is 0.722, against 0.767 for the depth-4 tree with 13 leaves, whose training accuracy is 0.843. So the big tree gains 13.5 points on the data it has seen and loses 4.5 on data it has not: the train/validation gap grows from about 8 points to about 26. Memorising 146 small groups is the high-variance end of the bias-variance trade.

**2. Refit the SVM search with only the linear kernel and compare its test ROC AUC with the RBF result.**

```python
linear_search = GridSearchCV(
    Pipeline([("scale", StandardScaler()), ("model", SVC(kernel="linear"))]),
    param_grid={"model__C": [0.1, 1, 10, 100]},
    cv=StratifiedKFold(5, shuffle=True, random_state=42),
    scoring="roc_auc",
    n_jobs=-1,
).fit(X_train, y_train)

comparison = pd.DataFrame({
    "best parameters": [search.best_params_, linear_search.best_params_],
    "CV ROC AUC": [search.best_score_, linear_search.best_score_],
    "test accuracy": [accuracy_score(y_test, search.predict(X_test)),
                      accuracy_score(y_test, linear_search.predict(X_test))],
    "test ROC AUC": [roc_auc_score(y_test, search.decision_function(X_test)),
                     roc_auc_score(y_test, linear_search.decision_function(X_test))],
}, index=["full search (linear + RBF)", "linear kernel only"])
print(comparison.round(4).to_string())
```

That is a tie, differing in the fourth decimal and in the opposite direction from cross-validation. So on these scaled features the RBF kernel buys nothing, and the simpler linear model (which also exposes one coefficient per feature) is the better choice on grounds of simplicity.

## Check your understanding

**1. What is the geometric meaning of a larger separating margin in an SVM?**

The margin is the gap between the decision boundary (a hyperplane in the kernel's feature space) and the nearest training points of each class, the support vectors; its total width is `2/‖w‖`. A wider margin means the boundary sits with more room on both sides, so a new point has to move further before it crosses to the wrong side. That is the SVM's built-in preference for the simplest, most stable boundary: maximising the margin is the same as minimising `‖w‖`, which limits how sharply the decision function can change. With overlapping classes a perfectly clean wide margin is impossible, so the soft-margin SVM trades margin width against violations, and `C` sets the exchange rate.

**2. How do `C` and, for an RBF kernel, `gamma` change the flexibility of the fitted decision boundary?**

- **`C`** is the penalty on margin violations. Large `C` punishes every misclassified or inside-the-margin point heavily, so the margin narrows and the boundary bends to fit individual points (flexible: low bias, high variance). Small `C` tolerates violations for a wider margin and a smoother boundary (less flexible: higher bias, lower variance).
- **`gamma`** sets how far each support vector's influence reaches in an RBF kernel. Large `gamma` means short reach, so the boundary forms tight islands around individual points (flexible, prone to overfitting). Small `gamma` means long reach and a very smooth boundary that approaches a linear one.

They interact: a large `gamma` with a large `C` is the most flexible combination, which is why in the grid above `gamma=0.1` was the weakest column and got worse at larger `C`, and why `C` and `gamma` are searched together instead of one at a time.

**3. Why should scaling and SVM hyperparameter selection both happen inside the validation procedure?**

The scaler learns means and standard deviations, and an SVM is built on distances and dot products, so the scale it sees changes the boundary. If the scaler is fitted on all rows before cross-validation, every validation fold has already influenced the statistics used to transform it: a small leak that makes scores look better than they will be on new data. Putting `StandardScaler` inside the `Pipeline` refits it on each training fold only. Hyperparameter selection has the same problem in a stronger form: this grid tries 24 combinations, and picking the best by its score on the same data you report would reward luck. Choosing `kernel`, `C` and `gamma` by cross-validation on the training partition only, and touching the test set once afterwards, keeps that final number an honest estimate.
